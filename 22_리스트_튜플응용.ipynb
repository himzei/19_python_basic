```python
# 반복문으로 리스트의 요소를 모두 출력하기 

a = [38, 21, 53, 62, 19]

for i in a:
    print(i)
```

```python
# 인덱스와 요소를 함께 출력하기 
a = [38, 21, 53, 62, 19]

for index, value in enumerate(a):
    print(index, value)

```

```python
a = [38, 21, 53, 62, 19]

for index, value in enumerate(a):
    print(index + 1, value)
```

```python
# while 반복문으로 요소 출력하기 
a = [38, 21, 53, 62, 19]

i = 0
while i < len(a):
    print(a[i])
    i += 1
```

```python
# 가장 작은 수와 가장 큰 수 구하기 
a = [38, 21, 53, 62, 19]

smallest = a[0]

for i in a:
    if i < smallest:
        smallest = i

print(smallest)
```

```python
# 가장 큰 수 구하기
a = [38, 21, 53, 62, 19]

largest = a[0]

for i in a:
    if i > largest:
        largest = i

print(largest)
```

```python
# 리스트 표현식 사용하기 
# [식 for 변수 in 리스트 ]

a = [i for i in range(10)]
a
```

```python
b = []

for i in range(10): 
    if i % 2 == 1:
        b.append(i + 5)

print(b)
```

```python
[i + 5 for i in range(10) if i % 2 == 1]
```

```python
[i * j for i in range(2, 10) for j in range(1, 10)]
```

```python
c = []
for i in range(2, 10):
    for j in range(1, 10):
        c.append(i * j)

print(c)
```
