```python
# 반복문으로 딕셔너리의 키-값 모두 출력하기 
x = {'a': 10, 'b': 20, 'c': 30, 'd': 40}

for i in x: 
    print(i)
```

```python
for i in x.items():
    print(i)
```

```python
for key, value in x.items():
    print(key, value)
```
