```python
# 불과 비교 연산자 사용하기 
print(True)
print(type(True))
print(False)
print(type(False))
```

```python
# 비교연산의 판단결과 
print(3 > 1)
print(1 > 3)
```

```python
# 숫자가 같은지 다른지 비교
print(10 == 10)
print(10 != 10)
```

```python
print(10 > 20)
print(10 < 20)
print(10 >= 10) 
print(10 <= 10)
```

```python
# 논리연산자 사용하기 
print(True and True)
print(True and False)
print(False and True)
print(False and False)
```

```python
# or 논리연산자
print(True or True)
print(True or False)
print(False or True)
print(False or False)
```

```python
# not x
print(not True)
print(not False)
```
