```python
# 파일객체 = open(파일이름, 파일모드)
# 파일객체.write(문자열)
# 파일객체.close()

file = open("hello.txt", "w")
file.write("hello world")
file.close()
```

```python
# 파일에서 문자열 읽기

file = open("hello.txt", "r")
s = file.read()
print(s)
file.close()
```

```python
# 자동으로 파일 객체 닫기
# with open("파일이름", "파일모드") as 파일객체: 
#     코드

with open("hello2.txt", "w") as file:
    for i in range(3): 
        file.write(f"Hello, world! {i}\n")
```
