```python
# 반복문으로 2차원 리스트의 요소를 모두 출력하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]

for x, y in a:
    print(x, y)
```

```python
# for 반복문을 두 번 사용하기 
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]

for i in a:
    for j in i:
        print(j, end=" ")
    print()
```

```python
# for와 range 사용하기 
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]

for i in range(len(a)):
    for j in range(len(a[i])):
        print(a[i][j], end=" ")
    print()
```

```python
# while 반복문을 한 번 사용하기
a = [
    [10, 20], 
    [30, 40], 
    [50, 60]
]

i = 0
while i < len(a):
    x, y = a[i]
    print(x, y)
    i += 1
```

```python
row, col = map(int, input().split())
```

```python
print(row)
print(col)
```

```python
matrix = []
for _ in range(row):
    matrix.append(list(input()))

```

```python
matrix
```

```python
for i in range(row):
    for j in range(col):
        if matrix[i][j] == "*": 
            print("*", end="")
        else:
            count = 0
            for r in range(i - 1, j + 2): 
                for c in range(i - 1, j + 2): 
                    if 0 <= r < row and 0 <= c < col:
                        if matrix[r][c] == "*":
                            count += 1 
            print(count, end="")
    print()
```

```python
col, row  = map(int,input().split())
matrix = []
for i in range(row):
    matrix.append(list(input()))

dx = [-1, -1, -1, 0, 0, 1, 1, 1]
dy = [-1, 0, 1, -1, 1, -1, 0, 1]
for i in range(row):
    for j in range(col):
        if matrix[i][j] != "*":
            count = 0
            for k in range(8):
                ni = i + dx[k]
                nj = j + dy[k]
                if 0 <= ni < row and 0 <= nj <col:
                    count += 1
            matrix[i][j] = str(count)
for i in range(row):
    print(''.join(matrix[i]))
```
